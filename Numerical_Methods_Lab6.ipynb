```python
import numpy as np
import scipy.linalg

# --- Task 3: Gaussian Elimination Implementation ---

def gaussian_elimination(A, b):
    n = len(b)
    # Create augmented matrix M
    M = np.zeros((n, n+1))
    for i in range(n):
        for j in range(n):
            M[i][j] = A[i][j]
        M[i][n] = b[i]

    # Pivoting and Elimination
    for k in range(n):
        # 1. Pivoting: Swap rows to put the largest absolute value on the diagonal
        for i in range(k, n):
            if abs(M[i][k]) > abs(M[k][k]):
                # Swap rows
                M[[k, i]] = M[[i, k]]

        # 2. Gaussian Elimination
        for j in range(k+1, n):
            if M[k][k] == 0:
                continue # Avoid division by zero
            q = float(M[j][k]) / M[k][k]
            for m in range(k, n+1):
                M[j][m] -= q * M[k][m]

    # 3. Back substitution
    x = [0 for _ in range(n)]
    x[n-1] = float(M[n-1][n]) / M[n-1][n-1]

    for i in range(n-1, -1, -1):
        z = 0
        for j in range(i+1, n):
            z = z + float(M[i][j]) * x[j]
        x[i] = float(M[i][n] - z) / M[i][i]

    return x

# Define the system of equations from Task 2
A = [[4, -1, 1],
     [2, 5, 2],
     [1, 2, 4]]
b = [8, 3, 11]

# Solve using our custom function
x_custom = gaussian_elimination(A, b)

# Solve using scipy library for comparison
x_scipy = scipy.linalg.solve(A, b)

# Output results
print("--- Task 3: Gaussian Elimination ---")
print("Solution using custom Gaussian Elimination:")
print(f"x1 = {x_custom[0]:.4f}, x2 = {x_custom[1]:.4f}, x3 = {x_custom[2]:.4f}")

print("\nSolution using scipy.linalg.solve:")
print(f"x1 = {x_scipy[0]:.4f}, x2 = {x_scipy[1]:.4f}, x3 = {x_scipy[2]:.4f}")
```

```python
# --- Task 4: LU Decomposition ---

def solve_lu(A, b):
    """
    Solves a linear system Ax = b using LU decomposition.
    """
    # Perform LU decomposition with pivoting
    # P is the permutation matrix, L is lower triangular, U is upper triangular
    P, L, U = scipy.linalg.lu(A)

    # Since A = P*L*U, the system is P*L*U*x = b
    # Multiplying both sides by P transpose gives: L*U*x = P^T * b

    # Step 1: Calculate modified b (P^T * b)
    b_prime = np.dot(P.T, b)

    # Step 2: Solve L * y = b_prime using forward substitution
    y = scipy.linalg.solve_triangular(L, b_prime, lower=True)

    # Step 3: Solve U * x = y using backward substitution
    x = scipy.linalg.solve_triangular(U, y, lower=False)

    return P, L, U, x

# Define the system of equations from the lecture
A2 = np.array([[1, 2, 1],
               [1, -2, 2],
               [2, 12, -2]])
b2 = np.array([0, 4, 4])

# Solve the system
P, L, U, x_lu = solve_lu(A2, b2)

# Output results
print("--- Task 4: LU Decomposition ---")
print("Matrix L (Lower triangular):")
print(L)
print("\nMatrix U (Upper triangular):")
print(U)
print("\nFinal Solution for [x1, x2, x3] using LU decomposition:")
print(f"x1 = {x_lu[0]:.4f}, x2 = {x_lu[1]:.4f}, x3 = {x_lu[2]:.4f}")
```
